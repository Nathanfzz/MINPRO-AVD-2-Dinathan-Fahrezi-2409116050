```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("top-500-movies.csv")
df
```

```python
penjualan_pergenre = df.groupby('genre')['worldwide_gross'].sum()
print(penjualan_pergenre)
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='genre', data=df,hue="genre", palette='pastel')
plt.title('jumlah film berdasarkan genre')
plt.xlabel('genre')
plt.ylabel('Jumlah film')
plt.xticks(rotation=45)
plt.show()
```

#Analisis dan saran

berdasarkan visualisasi diatas genre paling banyak adalah action dan adventure dikarenakan dua genre tersebut adalah genre yang bisa di terima semua kalangan karena seru dan menegangkan yang membuat penghasilan film dengan dua genre tersebut sangat tinggi. Para produsen film disarankan untuk membuat filmnya dengan genre action atau adventure atau setidaknya sebagai sub genre film mereka jika ingin membuat film romance atau horror misalnya agar lebih mudah untuk diterima pasar.

```python
plt.figure(figsize=(50, 40))
sns.scatterplot(x='worldwide_gross', y='theaters', hue='title', data=df)
plt.title('pendapatan dan theaters')
plt.xlabel('Kepuasan Pengunjung')
plt.ylabel('pendapatan')
plt.legend(title='genre')
plt.show()
```

#Analisis dan saran

Dengan melihat visualisasi diatas bisa dilihat bahwa judul yang terkenal dan besar mempengaruhi pendapatan dan rating. Disarankan untuk produsen film untuk mengadaptasi judul besar seperti avanger atau sejenisnya
